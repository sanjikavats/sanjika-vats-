```python
import random
numbers = [random.randint(16,69) for i in range (16)]

print("Random Numbers:", numbers)
print("Even Numbers:")
for num in numbers: 
    if num % 2 == 0:
        print(num)
```

```python
text = "hello everyone "
print("Quote:",text)
reverse_string = text[::-1]
print("Reverse Quote:", reverse_string)

reverse_words = " ".join(text.split()[::-1])
print("Reverse Words In Quote:", reverse_words)
```
